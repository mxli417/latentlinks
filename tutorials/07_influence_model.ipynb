# 07 — The Influence Model

For target `c`, source `j`, source state `r`, and target next state `s`, $A^{j\to c}_{rs}=P(z_{t+1}^{(c)}=s\mid z_t^{(j)}=r)$. Influence weights satisfy $D[c,j]\ge0$ and $\sum_jD[c,j]=1$. The target transition is $\sum_jD[c,j]A^{j\to c}_{z_t^{(j)},s}$.


```python
from latentlinks.influence import mit_six_chain_model

model = mit_six_chain_model()
observations, states = model.sample(12, seed=4)
print(model.influence)
print(states[:4])
print(observations[:4])
```

Read `D[target, source]` as source contribution to target dynamics. It is a predictive influence coefficient, not automatically a causal effect.
