Independent HMMs cannot express that one latent process changes another's dynamics. A general coupled HMM permits $P(z_{t+1}^{(c)}\mid z_t^{(1)},\ldots,z_t^{(C)})$. If each chain has $K$ states, the joint state has $K^C$ values. For 20 binary chains that is 1,048,576 states per time step.

A CHMM is a broad model family. The Influence Model introduced next is one structured parameterization; the names are not synonyms.
