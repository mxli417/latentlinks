# 02 — From every path to the forward algorithm

For $K=2,T=4$, only 16 state paths exist. We first compute $P(x)=\sum_z P(x,z)$ literally. Then define $\alpha_t(j)=P(x_{0:t},z_t=j)$ and reuse common prefixes: $\alpha_t(j)=B_j(x_t)\sum_i\alpha_{t-1}(i)A_{ij}$.


```python
import torch

from latentlinks.hmm import CategoricalHMM, brute_force_inference, forward_probabilities

model = CategoricalHMM([0.6, 0.4], [[0.7, 0.3], [0.2, 0.8]], [[0.9, 0.1], [0.25, 0.75]])
x = torch.tensor([0, 1, 1, 0])
oracle = brute_force_inference(model, x)
alpha = forward_probabilities(model, x)
print(alpha)
torch.testing.assert_close(oracle.likelihood, alpha[-1].sum())
```

The equality is the proof case. For long sequences, probabilities underflow; `forward_log_probabilities` applies the same recurrence with `torch.logsumexp`.
