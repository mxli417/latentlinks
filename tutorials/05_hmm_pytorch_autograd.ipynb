# 05 — Differentiating through HMM inference

Softmax maps unconstrained logits to $\pi$, $A$, and $B$. The log-space forward recursion remains the latent-variable inference algorithm. Autograd differentiates its observed-data likelihood; it does **not** replace inference.


```python
from latentlinks.hmm import CategoricalHMM, fit_gradient

source = CategoricalHMM([0.7, 0.3], [[0.9, 0.1], [0.2, 0.8]], [[0.8, 0.2], [0.1, 0.9]])
data = [source.sample(30, seed)[0] for seed in range(4)]
start = CategoricalHMM.random(2, 2, seed=5)
fitted, history = fit_gradient(start, data, iterations=50)
history.log_likelihoods[0], history.log_likelihoods[-1], fitted.transition
```

Compare this trace with EM. Gradient steps need not be monotone; learning rate matters. EM uses closed-form expected-count maximization for this model.
