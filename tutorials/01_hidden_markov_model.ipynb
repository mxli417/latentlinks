# 01 — A hidden Markov model

**Question.** How can an observed sequence be generated by an unobserved state process?

The hidden state $z_t$ is Markov: its future depends on $z_{t-1}$, not the full past. An observation $x_t$ depends on the current hidden state. The parameters are $\pi_k=P(z_0=k)$, $A_{ij}=P(z_t=j\mid z_{t-1}=i)$, and $B_{jm}=P(x_t=m\mid z_t=j)$.


```python
import torch

from latentlinks.hmm import CategoricalHMM

model = CategoricalHMM(
    initial=torch.tensor([0.6, 0.4]),
    transition=torch.tensor([[0.7, 0.3], [0.2, 0.8]]),
    emission=torch.tensor([[0.9, 0.1], [0.25, 0.75]]),
)
observations, states = model.sample(length=5, seed=3)
states, observations
```

## Check and next step

Rows are probability distributions. Sampling shows the generative direction $z_{t-1}\to z_t\to x_t$. Next we will reverse the question: given only `observations`, how likely are they?
