# 10 — The public MIT six-chain experiment

The usage page documents six binary latent chains, binary noisy observations, length 1000, transition `[[.99,.01],[.08,.92]]`, emission `[[.9,.1],[.1,.9]]`, and two equal-weight influence blocks. We reconstruct those parameters independently; no toolbox source is included.


```python
import torch

from latentlinks.influence import mit_six_chain_model

model = mit_six_chain_model()
x, z = model.sample(1000, seed=7)
assert x.shape == z.shape == (1000, 6)
torch.testing.assert_close(model.influence.sum(1), torch.ones(6, dtype=torch.float64))
model.influence
```

This verifies the documented setup, not behavioral equivalence to a particular toolbox version. `tests/reference/README.md` records provenance and the requirements for any future numeric fixture.
