# 08 — A factorized marginal approximation

Exact inference tracks dependence between all chains. Here we approximate the joint trajectory posterior by separate chain factors. Holding other chains' marginals fixed induces a time-varying effective transition for one target chain; we then reuse HMM forward–backward and update coordinates.

The discarded information is posterior dependence between chains. This project-derived approximation is not claimed to be Dong's nonlinear algorithm.


```python
from latentlinks.influence import factorized_inference, mit_six_chain_model

model = mit_six_chain_model()
observations, _ = model.sample(30, seed=2)
result = factorized_inference(model, observations, max_iterations=50)
print(result.max_changes)
print(result.marginals.shape)  # chains, time, states
print(result.marginals.sum(dim=2))
```

The monitored statistic is maximum marginal change, not a proven ELBO. Convergence of that number does not establish correctness of the approximation. Independent-chain tests compare it with exact inference.
