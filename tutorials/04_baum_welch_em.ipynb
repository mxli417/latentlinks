# 04 — Baum–Welch is expected counting

If states were visible, transitions could be counted. EM replaces missing counts by $N_{ij}=E[\#(i\to j)\mid X,\theta]=\sum_t\xi_t(i,j)$. The M-step normalizes expected counts row by row. Every denominator is the expected number of opportunities for that event.


```python
from latentlinks.hmm import (
    CategoricalHMM,
    baum_welch_step,
    forward_log_likelihood,
)

source = CategoricalHMM([0.8, 0.2], [[0.9, 0.1], [0.15, 0.85]], [[0.9, 0.1], [0.1, 0.9]])
data = [source.sample(50, seed)[0] for seed in range(6)]
model = CategoricalHMM.random(2, 2, seed=11)
before = sum(forward_log_likelihood(model, x) for x in data)
model, _ = baum_welch_step(model, data)
after = sum(forward_log_likelihood(model, x) for x in data)
assert after >= before - 1e-8
before.item(), after.item(), model.transition, model.emission
```

Exact EM should not materially reduce observed-data likelihood. It can still reach a local optimum, and state labels may be permuted without changing the model distribution.
