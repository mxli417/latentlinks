# 03 — Backward messages and posterior states

$\beta_t(i)=P(x_{t+1:T}\mid z_t=i)$. Combining forward and backward evidence gives $\gamma_t(i)=P(z_t=i\mid x)$ and $\xi_t(i,j)=P(z_t=i,z_{t+1}=j\mid x)$. `gamma` answers *which state generated this observation?*; `xi` answers *which transition occurred here?*


```python
import torch

from latentlinks.hmm import CategoricalHMM, brute_force_inference, forward_backward

model = CategoricalHMM([0.6, 0.4], [[0.7, 0.3], [0.2, 0.8]], [[0.9, 0.1], [0.25, 0.75]])
x = torch.tensor([0, 1, 1, 0])
result = forward_backward(model, x)
oracle = brute_force_inference(model, x)
torch.testing.assert_close(result.gamma.sum(1), torch.ones(4, dtype=torch.float64))
torch.testing.assert_close(result.xi.sum((1, 2)), torch.ones(3, dtype=torch.float64))
torch.testing.assert_close(result.gamma, oracle.gamma)
result.gamma, result.xi
```

Normalization is not cosmetic: it states that each posterior is a probability distribution. These posterior probabilities become expected counts in EM.
